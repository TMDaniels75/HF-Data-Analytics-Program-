```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
import glob
```

```python
df=pd.read_csv("https://raw.githubusercontent.com/niteen11/DataAnalyticsAcademy/master/Python/dataset_diabetes/diabetic_data.csv")
```

```python
df.head()
```

```python
df.columns
```

```python
df.shape
```

```python
df['gender'].value_counts()
```

```python
df = df[df['gender'] != 'Unknown/Invalid']
```

```python
df['gender'].value_counts()
```

```python
male = df[df['gender'] == 'Male']['num_medications']

female = df[df['gender'] == 'Female']['num_medications']
```

```python
stats = pd.DataFrame({
    'Male': male.describe(),
    'Female': female.describe()
})

stats
```

```python
avg_meds = df.groupby('gender')['num_medications'].mean()

avg_meds.plot(
    kind='bar',
    figsize=(8,5)
)

plt.title('Average Number of Medications by Gender')
plt.xlabel('Gender')
plt.ylabel('Average Number of Medications')
plt.xticks(rotation=0)

plt.show()
```
